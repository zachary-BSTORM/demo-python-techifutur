# Les collections

Python propose plusieurs types de collections pour stocker des groupes de valeurs. Chaque type a ses caractéristiques propres.

| Type | Syntaxe | Ordonné | Modifiable | Doublons | Accès |
|---|---|---|---|---|---|
| **Liste** | `[1, 2, 3]` | ✅ | ✅ | ✅ | Index |
| **Tuple** | `(1, 2, 3)` | ✅ | ❌ | ✅ | Index |
| **Set** | `{1, 2, 3}` | ❌ | ✅ | ❌ | Itération |
| **Dictionnaire** | `{"a": 1}` | ✅ (3.7+) | ✅ | Clés uniques | Clé |

> **Vitesse d'accès** : la **complexité** mesure comment le temps d'une opération évolue quand la collection grossit. Les listes et tuples sont contigus en mémoire, donc le test `in` doit parcourir les éléments un par un → `O(n)` (le temps grandit avec la taille). Les sets et dicts utilisent une table de hachage (hash table) : le test `in` se fait en `O(1)`, c'est-à-dire en temps quasi constant, quelle que soit la taille.

```python
# List

ma_liste = [1,2,3,4,5]
print(ma_liste[0]) # 1
print(ma_liste[-1]) # 5
print(ma_liste)
```

### Méthodes de la liste — référence complète

| Méthode | Signature | Description | Retourne |
|---|---|---|---|
| `append` | `append(x)` | Ajoute `x` à la fin | `None` |
| `extend` | `extend(iterable)` | Ajoute tous les éléments de l'itérable | `None` |
| `insert` | `insert(i, x)` | Insère `x` à l'index `i` | `None` |
| `remove` | `remove(x)` | Supprime la **première** occurrence de `x` (ValueError si absent) | `None` |
| `pop` | `pop(i=-1)` | Supprime et retourne l'élément à l'index `i` (dernier par défaut) | l'élément |
| `clear` | `clear()` | Vide la liste | `None` |
| `index` | `index(x, start=0, stop=len)` | Index de la 1ère occurrence de `x` entre `start` et `stop` | `int` |
| `count` | `count(x)` | Nombre d'occurrences de `x` | `int` |
| `sort` | `sort(key=None, reverse=False)` | Trie la liste **en place** | `None` |
| `reverse` | `reverse()` | Inverse la liste **en place** | `None` |
| `copy` | `copy()` | Copie superficielle (shallow copy) | `list` |

> **`sort()` vs `sorted()`** : `sort()` modifie la liste en place et retourne `None`. `sorted(iterable, key=None, reverse=False)` retourne une **nouvelle** liste triée sans modifier l'original.

```python
new_liste = [1,2,6,3,4,5,6]
    # index [0,1,2,3,4,5,6]
    # commence à 0
    # termine à taille -1

# ajout 
new_liste.append(7)
print("append")
print(new_liste)

# ajout à un emplacement
new_liste.extend([8,9,10])
print("extend")
print(new_liste)

# retirer un élément : uniquement la première occurence
new_liste.remove(6)
print("remove")
print(new_liste)

# supprime et renvoie l'élement
deleted = new_liste.pop()
print("pop")
print(new_liste)
print(deleted)

# supprime et renvoie l'élement à une position
deleted = new_liste.pop(0)
print("pop")
print(new_liste)
print(deleted)

# accès par index
liste_index = [10,20,30,40]
print("index")
print(liste_index.index(20))

# accès par index à partir d'un index
liste_index = [10,20,30,40,20,50,60]
print("index avec position")
print(liste_index.index(20,2))

# taille du tableau
print("taille list")
print(len(liste_index))

# accès au dernier par la taille
print("accès taille list - 1")
test =  liste_index[       7         - 1]
value = liste_index[len(liste_index) - 1]
print(value)

# trie de la liste sans paramètres
mots = ["kiwi","banane", "pomme","abricot"]
mots.sort() # tri alphabetique
print("tri sort sans params")
print(mots)

# trie avec paramètre
mots.sort(reverse=True) # tri alphabetique inversé
print("sort avec params")
print(mots)

# trie sur base d'un critère
mots.sort(key=len) # tri par rapport à la taille du mot
print("sort len")
print(mots)

mots_2 = ["kiwi","banane", "pomme","abricot"]
# trie sur base d'un critère 2
mots_2.sort(key=lambda m:m[-1] ) # tri par rapport à la dernière lettre
print("sort avec lambda")
print(mots_2)

# retourne le nombre d'éléments présent
liste_multiple_value = [1,2,3,1,2]
print("count")
print(liste_multiple_value.count(1))

# vider la liste
liste_multiple_value.clear()
print("clear")
print(liste_multiple_value)
```

## Les tuples (`tuple`)

Le tuple est similaire à la liste, mais **immutable** (non modifiable après création).

- **Indexé** 
- **Ordonné** : l'ordre est conservé
- **Immutable** : on ne peut pas modifier, ajouter ou supprimer des éléments
- **Accepte les doublons**

```python
# Création
mon_tuple = (1, 2, 3)
mon_tuple = 1, 2, 3          # les parenthèses sont optionnelles
singleton = (42,)            # ATTENTION : la virgule est obligatoire pour un tuple à 1 élément
pas_tuple = (42)             # ceci est juste un int entre parenthèses
mon_tuple = tuple([1, 2, 3]) # depuis une liste
mon_tuple = tuple("abc")     # depuis une chaîne → ('a', 'b', 'c')

# Accès
print(mon_tuple[0])   # Premier élément
print(mon_tuple[-1])  # Dernier élément

# Unpacking (déballage) — très utilisé en Python
a, b, c = (1, 2, 3)
premier, *reste = (1, 2, 3, 4)  # reste = [2, 3, 4]
*debut, dernier = (1, 2, 3, 4)  # debut = [1, 2, 3]

# Retour multiple depuis une fonction
def coordonnees():
    return 10, 20  # retourne un tuple implicitement

x, y = coordonnees()

tuple_1 = tuple([2,1,3,4])
print(tuple_1)
```

### Méthodes du tuple — référence complète

Le tuple n'a que **2 méthodes** (il est immutable) :

| Méthode | Signature | Description | Retourne |
|---|---|---|---|
| `count` | `count(x)` | Nombre d'occurrences de `x` | `int` |
| `index` | `index(x, start=0, stop=len)` | Index de la 1ère occurrence de `x` entre `start` et `stop` (ValueError si absent) | `int` |

### Fonctions built-in applicables aux tuples

| Fonction | Signature | Description |
|---|---|---|
| `len` | `len(t)` | Nombre d'éléments |
| `max` | `max(t, key=None)` | Valeur maximale |
| `min` | `min(t, key=None)` | Valeur minimale |
| `sum` | `sum(t, start=0)` | Somme des éléments (+ `start`) |
| `sorted` | `sorted(t, key=None, reverse=False)` | Nouvelle **liste** triée |
| `reversed` | `reversed(t)` | Itérateur inversé |
| `enumerate` | `enumerate(t, start=0)` | Itérateur de `(index, valeur)` |
| `zip` | `zip(t1, t2, ...)` | Itérateur de tuples appairés |

```python
mon_tuple = (10, 20, 30, 20, 10)

# count(x)
print(mon_tuple.count(20))      # 2

# index(x, start, stop)
print(mon_tuple.index(20))      # 1
print(mon_tuple.index(20, 2))   # 3 (cherche à partir de l'index 2)
print(mon_tuple.index(20, 2,5))   # 3 (cherche à partir de l'index 2 jusqu'a l'index 5)

# len, max, min, sum
print(len(mon_tuple))           # 5
print(max(mon_tuple))           # 30
print(min(mon_tuple))           # 10
print(sum(mon_tuple))           # 90
print(sum(mon_tuple, 100))      # 190 (100 + somme)

# sorted → retourne une liste, pas un tuple
liste_triee = sorted(mon_tuple)  # [10, 10, 20, 20, 30]

# reversed → itérateur, convertir en tuple
inverse = tuple(reversed(mon_tuple))  # (10, 20, 30, 20, 10)

# enumerate
for i, val in enumerate(mon_tuple, start=1):
    print(f"Index {i} → {val}")

# zip
noms = ("Alice", "Bob", "Charlie")
ages = (25, 30, 35)
for nom, age in zip(noms, ages):
    print(f"{nom} a {age} ans")
```

## Les ensembles (`set`)

Le set est une collection de **valeurs uniques et non ordonnées**.

- **Non indexé** : pas d'accès par indice (`[0]`)
- **Non ordonné** : l'ordre d'itération n'est pas garanti
- **Mutable** : on peut ajouter ou supprimer des éléments
- **Pas de doublons** : les valeurs dupliquées sont automatiquement ignorées
- **Optimisé** pour tester l'appartenance avec `in` → O(1)
- Les éléments doivent être **hashables** (pas de liste ni de dict comme élément)
- `frozenset` : version **immutable** du set (hashable, utilisable comme clé de dict)

### Méthodes du set — référence complète

#### Modification
| Méthode | Signature | Description | Retourne |
|---|---|---|---|
| `add` | `add(elem)` | Ajoute `elem` (sans effet s'il est déjà présent) | `None` |
| `remove` | `remove(elem)` | Supprime `elem` (KeyError si absent) | `None` |
| `discard` | `discard(elem)` | Supprime `elem` **sans erreur** s'il est absent | `None` |
| `pop` | `pop()` | Supprime et retourne un élément **arbitraire** (KeyError si vide) | élément |
| `clear` | `clear()` | Vide le set | `None` |
| `copy` | `copy()` | Copie superficielle | `set` |
| `update` | `update(*iterables)` | Ajoute tous les éléments de plusieurs itérables | `None` |

#### Opérations ensemblistes
| Méthode | Opérateur | Signature | Description | Retourne |
|---|---|---|---|---|
| `union` | `\|` | `union(*sets)` | Union de plusieurs sets | `set` |
| `intersection` | `&` | `intersection(*sets)` | Intersection | `set` |
| `difference` | `-` | `difference(*sets)` | Différence (éléments dans self mais pas dans others) | `set` |
| `symmetric_difference` | `^` | `symmetric_difference(other)` | Éléments dans l'un ou l'autre, pas les deux | `set` |
| `issubset` | `<=` | `issubset(other)` | Vérifie si self ⊆ other | `bool` |
| `issuperset` | `>=` | `issuperset(other)` | Vérifie si self ⊇ other | `bool` |
| `isdisjoint` | — | `isdisjoint(other)` | Vérifie si self ∩ other = ∅ | `bool` |

> **Variantes `_update`** : `intersection_update()`, `difference_update()`, `symmetric_difference_update()` modifient le set **en place** et retournent `None`.

```python
# Création
my_set = {1, 2, 3}
my_set = set([1, 2, 2, 3])   # doublons supprimés → {1, 2, 3}
my_set = set("hello")         # {'h', 'e', 'l', 'o'} (l en un seul exemplaire)
my_set = set()                # set vide — JAMAIS {} (ça crée un dict !)

# add vs update
my_set = {1, 2, 3}
my_set.add(4)                 # ajoute un seul élément
my_set.update([5, 6], {7})    # ajoute plusieurs itérables d'un coup

# remove vs discard
my_set.remove(1)              # KeyError si 1 absent
my_set.discard(999)           # silencieux si 999 absent

# Opérations ensemblistes
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)            # Union          → {1, 2, 3, 4, 5, 6}
print(A.union(B))       # identique

print(A & B)            # Intersection   → {3, 4}
print(A.intersection(B))

print(A - B)            # Différence     → {1, 2}
print(A.difference(B))

print(A ^ B)            # Sym. diff.     → {1, 2, 5, 6}
print(A.symmetric_difference(B))

# Tests ensemblistes
print({1, 2}.issubset({1, 2, 3}))    # True  → {1,2} ⊆ {1,2,3}
print({1, 2, 3}.issuperset({2, 3}))  # True
print({1, 2}.isdisjoint({3, 4}))     # True  → aucun élément commun

# frozenset — version immutable
fs = frozenset([1, 2, 3])
# fs.add(4)  # AttributeError — non modifiable
d = {fs: "valeur"}  # utilisable comme clé de dict
```

## Les dictionnaires (`dict`)

Le dictionnaire stocke des paires **clé : valeur**.

- **Non indexé par position** : on accède aux valeurs via leur clé
- **Ordonné** depuis Python 3.7 (l'ordre d'insertion est conservé)
- **Mutable** : on peut ajouter, modifier ou supprimer des paires
- **Clés uniques** : une clé dupliquée écrase la valeur précédente
- **Accès rapide** aux valeurs : O(1) grâce aux tables de hachage
- Les **clés** doivent être **hashables** (str, int, float, tuple...)
- Les **valeurs** peuvent être de n'importe quel type

### Méthodes du dictionnaire — référence complète

| Méthode | Signature | Description | Retourne |
|---|---|---|---|
| `get` | `get(key, default=None)` | Valeur associée à `key`, ou `default` si absent | valeur ou `None` |
| `keys` | `keys()` | Vue sur toutes les clés (dynamique) | `dict_keys` |
| `values` | `values()` | Vue sur toutes les valeurs (dynamique) | `dict_values` |
| `items` | `items()` | Vue sur toutes les paires `(clé, valeur)` | `dict_items` |
| `pop` | `pop(key, default=...)` | Supprime `key` et retourne sa valeur (KeyError si absent et pas de default) | valeur |
| `popitem` | `popitem()` | Supprime et retourne la **dernière** paire insérée (LIFO) | `(clé, valeur)` |
| `update` | `update(other=None, **kwargs)` | Fusionne avec un autre dict ou des kwargs | `None` |
| `setdefault` | `setdefault(key, default=None)` | Retourne la valeur de `key`; si absente, l'insère avec `default` | valeur |
| `clear` | `clear()` | Vide le dictionnaire | `None` |
| `copy` | `copy()` | Copie superficielle | `dict` |
| `fromkeys` | `dict.fromkeys(iterable, value=None)` | Crée un dict avec les clés de l'itérable (méthode de classe) | `dict` |

> **Vue dynamique** : `keys()`, `values()`, `items()` retournent des **vues** qui reflètent les changements du dict en temps réel. Pour une copie statique, utilisez `list(d.keys())`.

```python
# Création
mon_dict = {"nom": "Alice", "age": 30}
mon_dict = dict(nom="Alice", age=30)          # avec kwargs
mon_dict = dict([("nom", "Alice"), ("age", 30)])  # depuis liste de tuples

# Dict comprehension
carres = {x: x**2 for x in range(5)}          # {0:0, 1:1, 2:4, 3:9, 4:16}
filtres = {k: v for k, v in mon_dict.items() if isinstance(v, str)}

# fromkeys — crée un dict avec des clés et une valeur par défaut
cles = ["a", "b", "c"]
d = dict.fromkeys(cles, 0)     # {'a': 0, 'b': 0, 'c': 0}
d = dict.fromkeys(cles)        # {'a': None, 'b': None, 'c': None}

# Accès
mon_dict = {"a": 1, "b": 2, "c": 3}
print(mon_dict["a"])           # 1 (KeyError si absent)
print(mon_dict.get("z"))       # None
print(mon_dict.get("z", -1))   # -1 (valeur par défaut)

# setdefault — utile pour initialiser des valeurs manquantes
compteur = {}
for lettre in "banana":
    compteur.setdefault(lettre, 0)
    compteur[lettre] += 1
# Résultat : {'b': 1, 'a': 3, 'n': 2}

# update — fusionne plusieurs sources
d1 = {"a": 1}
d2 = {"b": 2, "c": 3}
d1.update(d2)                  # {'a': 1, 'b': 2, 'c': 3}
d1.update(d=4, e=5)            # peut aussi prendre des kwargs

# Fusion Python 3.9+ avec l'opérateur |
d3 = d1 | d2                   # retourne un nouveau dict fusionné
d1 |= d2                       # équivalent de update, en place

# pop(key, default)
val = mon_dict.pop("a")        # retourne 1, supprime 'a'
val = mon_dict.pop("z", -1)    # retourne -1, pas de KeyError

# popitem() — dernier inséré (utile pour traiter un dict comme une pile)
cle, valeur = mon_dict.popitem()

# Vues (views)
mon_dict = {"x": 10, "y": 20}
cles    = mon_dict.keys()      # dict_keys(['x', 'y'])
valeurs = mon_dict.values()    # dict_values([10, 20])
paires  = mon_dict.items()     # dict_items([('x',10), ('y',20)])

# Les vues sont dynamiques
mon_dict["z"] = 30
print(list(cles))              # ['x', 'y', 'z'] — mis à jour automatiquement

# Parcours
for cle in mon_dict:                         # itère sur les clés
    print(cle, mon_dict[cle])

for cle, valeur in mon_dict.items():         # itère sur les paires
    print(f"{cle} = {valeur}")
    
for value in mon_dict.values(): # itère sur les valeurs
    print(f"Value : {value}")

# Tester l'existence d'une clé
print("x" in mon_dict)         # True
print(10 in mon_dict.values()) # True

# Copie
import copy
copie_superficielle = mon_dict.copy()        # ou dict(mon_dict)
copie_profonde = copy.deepcopy(mon_dict)     # pour les dicts imbriqués
```

# Les itérations (boucles)

Les boucles permettent de **répéter** des instructions un certain nombre de fois ou tant qu'une condition est remplie.

La condition doit toujours renvoyer un **booléen** (`True` ou `False`).

Python propose deux types de boucles :
- **`while`** : répète tant qu'une condition est vraie
- **`for`** : parcourt une séquence d'éléments

```python
compteur = 0

while compteur < 10:
    print(compteur)
    
victory = False

while not victory:
    info = input("Voulez-vous continuez ? y - n")
    
    if info == "y":
        victory = False
```

For

```python
collection = [1,2,3,4]

for value in collection:
    print(value)
```

### Paramètres de `range()`

| Paramètre | Description | Valeur par défaut |
|---|---|---|
| `start` | Premier entier de la séquence | `0` |
| `stop` | Dernier entier (**non inclus**) | — |
| `step` | Pas entre chaque entier | `1` |

```python
for i in range(10):
    print(i)
    
for i in range(5 , 10 , 2):
    print(i)
```

```python
collection_value = ["a","b","c"]

for i,value in enumerate(collection_value):
    print(f" la valeur est {value} en position {i}")
```

## Contrôle de boucle : `break` et `continue`

| Mot-clé | Description |
|---|---|
| `break` | **Interrompt** la boucle immédiatement |
| `continue` | **Saute** l'itération en cours et passe à la suivante |

```python
# break : arrête la boucle quand on trouve 3
for i in range(10):
    if i == 3:
        break
    print(i)  # Affiche 0, 1, 2

print("---")

# continue : saute les nombres pairs
for i in range(6):
    if i % 2 == 0:
        continue
    print(i)  # Affiche 1, 3, 5
```

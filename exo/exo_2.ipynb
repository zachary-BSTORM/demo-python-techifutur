#### Calcul moyenne de température

- Calculez la somme et la moyenne des valeurs d'une liste,


```python
temperatures = [18.5, 21.0, 19.75, 23.5, 17.25, 20.0, 22.5]

somme = 0

for v in temperatures:
    somme += v
    
moyenne = somme / len(temperatures)

print(moyenne)

moyenne = round(moyenne,2)

print(moyenne)
```

#### Inversion d'une liste

- Inverser une liste Construisez une nouvelle liste contenant les éléments de la liste d'origine dans l'ordre inverse


```python
langages = ["Python", "C#", "JavaScript", "SQL", "TypeScript"]

# avec slicing
inverse = langages[::-1]

print(langages)
print(inverse)

langages.reverse()

inverse_reverse = langages[:]

# avec reverse
langages.reverse()

print(langages)
print(inverse_reverse)
```

#### Fusion de dictionnaire 

Fusionner deux stocks Deux entrepôts envoient leur inventaire. 
Construisez un dictionnaire du stock total : 
les produits communs voient leurs quantités additionnées, 
les produits présents dans un seul entrepôt sont repris tels quels.

```python
entrepot_a = {"clavier": 12, "souris": 30, "ecran": 5}
entrepot_b = {"souris": 15, "ecran": 8, "webcam": 7, "casque": 3}

entrepot_c = entrepot_a

for produit,quantity in entrepot_b.items():
    if produit in entrepot_c:
        entrepot_c[produit] += quantity
    else:
        entrepot_c[produit] = quantity
        
for k,v in entrepot_c.items():
    print(f"nom : {k} : quantity : {v}")
```

```python
# Programme interactif : ! ne pas exécuter dans jupyter notebook !
"""
Le panier de courses Écrivez un programme avec un menu qui boucle 
jusqu'à ce que l'utilisateur choisisse de quitter.

1. Ajouter un produit
2. Retirer un produit
3. Afficher le panier
4. Vider le panier
5. Quitter
Votre choix :
"""

menu = """
1. Ajouter un produit
2. Retirer un produit
3. Afficher le panier
4. Vider le panier
5. Quitter
"""
panier = {}

stop_app = False

while not stop_app:
    print("==========================================")
    print(menu)
    
    choice = input("Entrez votre choix : ")
    
    match choice:
        case "1":
            key = input("Entrez l'identifiant du produit : ")
            
            name = input("Entrez le nom du produit : ")
            price = float(input("Entrez le prix du produit : "))
            
            panier[key] = {"name":name,"price":price}
            
            
        case "2":
            key = input("Entrez l'identifiant du produit à supprimer : ")
            
            if key in panier.keys():
                del panier[key]
            else:
                print("ce produit n'est pas dans le stock!")
                
                
        case "3":
            for key,value in panier.items():
                print(f"{key} : name : {value["name"]} - price : {value["price"]}€")
            
        case "4":
            panier.clear()
            
        case "5":
            stop_app = True
            
    
    if stop_app == True:
        print("Merci à bientot !")
```

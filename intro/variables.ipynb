## <center>Les types de base</center>

| Type | Nom complet | Exemple | Utilité |
|---|---|---|---|
| `int` | entier | `42` | compter, indexer |
| `float` | nombre à virgule | `3.14` | mesures, prix |
| `str` | chaîne de caractères | `"bonjour"` | texte |
| `bool` | booléen | `True` / `False` | vrai ou faux |

- La fonction `type()` permet de demander à Python le type d'une valeur

type() : permet de récupérer le type de la variable

```python
age = 20

print(type(age))

age = "hello"

print(type(age))
```

Afficher ou récupérer une information

afficher dans la console : print("")

récupération depuis la console en type str : input("entrez une valeur : ")

```python
print("Entrez vos informations")

nom = input("Entrez votre nom")

print(nom)
print(type(nom))

```

Convertir une valeur

int() : conversion en entier
float() : conversion en float
str() : conversion en chaine de caractères
bool() : conversion en booléen

```python
age = input("Entrez votre age : ")

age_entier = int(age)

print(type(age_entier))
print(age + 1)
```

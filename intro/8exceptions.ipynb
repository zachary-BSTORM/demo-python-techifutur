## <center>`try` / `except` — le filet de sécurité</center>
- `try` : « essaie ce bloc »
- `except` : « si telle exception se produit, fais ceci à la place » — le programme **continue**

```python

def get_number():
    while True:
        try:
            value = None
            
            value = int(input("Entrez un nombre : "))
            
            if type(value) is int:
                return value
        except:
            print("Veuillez recommencer !")
            
```

```python


try:
    value = int(input("Entrez un nombre"))
except:
    print("La valeur est incorrect")
else:
    print(f"Valeur : {value}")
finally:
    print("Fin d'exécution")
```

```python

ma_liste = [1,2,3,4,5,6]

try:
    
    index = int(input("Entrez l'index de l'élément à modifier : "))
    new_value = int(input("Entrez la nouvelle valeur : "))
    ma_liste[index] = new_value
    
except ValueError:
    print("La valeur est incorrect")
except IndexError:
    print("L'index n'existe pas dans la liste")
except:
    print("autre erreur")
else:
    for v in ma_liste:
        print(v)
```

Déclencher une erreur
- raise Error("")

```python
value = 5

if value == 5:
    raise ValueError("La valeur est incorrecte")
```

```python
try:
    value = 5

    if value == 5:
        raise ValueError("La valeur est incorrecte")
    
except ValueError as e:
    print(f"Une erreur est survenue : {e}")
    
```

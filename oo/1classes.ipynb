## <center>L'idée de la Programmation Orientée Objet</center>
- Jusqu'ici, nos données (dicts) et nos fonctions vivent **séparément**
- La POO les **regroupe** : une **classe** est un moule qui définit à la fois les données et les actions qui vont avec
- Un **objet** est un exemplaire fabriqué avec ce moule

| Concept | Analogie | Exemple |
|---|---|---|
| Classe | le moule à gaufres | `Tache` |
| Objet (ou *instance*) | une gaufre | « Préparer la réunion » |
| Attributs | les caractéristiques | le titre, faite ou non |
| Méthodes | les actions | terminer la tâche, l'afficher |

- Vous utilisez déjà des objets sans le savoir : `"texte".upper()`, `liste.append(...)` — tout est objet en Python !

```python
class User():
    def __init__(self,name_params,email_params,password_params):
        self.name = name_params
        self.email = email_params
        self.password = password_params

    
    def se_presenter(self,animal):
        print(f"L'utilisateur : {self.name} - email : {self.email} - password : {self.password} : possède l'animal : {animal}")

        
        
        


user_1 = User("bob","mail@mail.com","admin1234")
user_2 = User("john","mail@mail.com")
```

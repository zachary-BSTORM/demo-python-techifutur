# Les fonctions et procédures

Une **fonction** permet d'encapsuler un bloc de code réutilisable.  
Au lieu de réécrire la même logique 10 fois, on l'écrit une fois dans une fonction et on l'appelle quand on en a besoin.

On déclare une fonction avec le mot-clé **`def`**.

| Type | Description | Mot-clé |
|---|---|---|
| **Fonction** | Renvoie une valeur | `return` |
| **Procédure** | Effectue une action, ne renvoie rien | Pas de `return` |

> En Python, toute fonction sans `return` retourne automatiquement `None` (= rien, vide).

````python
def methode_simple():
    print("hello world")
    
def methode_return():
    return "hello world return"

methode_simple()


result = methode_return()
print(result)
````

## Paramètres et valeurs par défaut

Une fonction peut recevoir des **paramètres** (des valeurs qu'on lui passe au moment de l'appel).  
On peut leur donner une **valeur par défaut** : si l'appelant ne fournit pas ce paramètre, la valeur par défaut est utilisée.

### Règle importante
Les paramètres **avec valeur par défaut** doivent toujours être placés **après** les paramètres obligatoires.

````python
def presentation(name):
    print(f"Bonjour je suis {name}")
    
def presentation_2(name,age = 30):
    print(f"bonjour je suis {name} et j'ai {age}ans")
    
presentation("bob")

presentation_2("bob")
presentation_2("bob",36)
````

Attention au valeur par défaut pour les collection 
- utiliser None plutot qu'une liste par défaut

````python


def afficher_liste(value,liste = None):
    if liste is None:
        liste= []

    liste.append(value)
    return liste

print(afficher_liste(1)) # [1]
print(afficher_liste(2)) # [1,2]
````

Globale

- Fonctionnement par défaut : copie la valeur
- Pour utiliser une variable globale

````python
# variable locale
value = 5

def increment(v):
    v += 1
    print(v)
    
increment(value) # 5 => 6

print(value) # 5
````

````python
# utilisation de valeur 
value_1 = 5

def decrement():
    value_1 = 6
    print(value_1)
    
decrement()
print(value_1)

# utilisation de la référence d'une variable
value_2 = 5

def increment():
    global value_2
    
    value_2 += 1
    
    print(value_2)
    
increment() # 5 => 6

print(value_2) # 6
````

## `*args` et `**kwargs` — Nombre variable d'arguments

Parfois on veut qu'une fonction accepte **autant d'arguments qu'on veut**, sans savoir à l'avance combien il y en aura.

| Syntaxe | Ce que ça reçoit | Type reçu dans la fonction |
|---|---|---|
| `*args` | Plusieurs arguments positionnels (sans nom) | `tuple` |
| `**kwargs` | Plusieurs arguments nommés (`clé=valeur`) | `dict` |

> **`args`** et **`kwargs`** sont juste des conventions de nommage — on pourrait les appeler `*nombres` ou `**options`.  
> Mais tout le monde utilise `*args` / `**kwargs`, donc reste sur ces noms.

**L'ordre obligatoire dans une signature :**
```
def f(obligatoires, *args, keyword_only, **kwargs)
```

````python
def affiche_params(name,firstname = "inconnu",*args):
    print(f"Bonjour je suis {name} - {firstname}")
    
    for v in args:
        print(f"valeur : {v}")
        
affiche_params("bob","doe",5,6,7,8,9)
affiche_params("bob","doe",5,6)
````

````python
def affiche_params(name,firstname = "inconnu",**kwargs):
    print(f"Bonjour je suis {name} - {firstname}")
        
    for k,v in kwargs.items():
        print(f" key : {k} - Value : {v}")
        
affiche_params("bob","doe",animal="garfield")
affiche_params("bob","doe",plat="lasagne",boisson="coca")
````

````python
def inscription(name,matiere,niveau):
    print(f"nom : {name} - matière : {matiere} - niveau : {niveau}")
    
inscription(matiere="python",name="bob",niveau=3)

inscription("python","bob",3)
````

````python
age = 30

def verifie_age(age):
    if age < 18:
        return
    print("Bienvenu ")
    
verifie_age(age)
verifie_age(16) # ici l'instruction se termine prématurément
````

````python
list_note = [10,15,18,3]

def return_min_max(collection):
    
    minimum = min(collection)
    maximum = max(collection)

    return minimum,maximum


result_tuple = return_min_max(list_note)

print(result_tuple)

a , b = return_min_max(list_note)

print(a)
print(b)
````

## Fonctions lambda

Une **lambda** est une fonction anonyme (sans nom) écrite en une seule ligne.  
Elle est utile pour des petites opérations simples qu'on passe en argument à une autre fonction.

```python
lambda paramètres : expression
```

> Une lambda ne peut contenir qu'**une seule expression** — pas de boucle, pas de bloc `if/else` **multiligne**.  
> En revanche, le `if/else` **en une ligne** (appelé **expression ternaire** : une valeur *si* condition *sinon* une autre valeur) est permis, car c'est une expression :
> ```python
> parite = lambda x: "pair" if x % 2 == 0 else "impair"
> ```
> Si la logique est complexe, écris une vraie fonction avec `def`.

````python
def carre(x):
    return x ** 2

carre_lambda = lambda x : x ** 2

print(carre(5))

print(carre_lambda(5))
````

````python
mots =  ["Python", "java", "C", "rust"]

trie = sorted(mots,key=lambda m : m.lower())

trie_normal = sorted(mots)

print(trie)
print(trie_normal)
````

````python
nombres = [1, 2, 3, 4, 5]

carres = list(map(lambda x : x ** 2 , nombres))

print(carres)
````

````python

def addition(a : int,b:float):
    """
    a : int
    b : float
    
    *args : ...
    
    **kwargs : ...
    
    renvoie la valeur de a + b
    
    """
    return a + b

result = addition("a","b")

result_int = addition(5,8.5)

print(result)
print(result_int)

help(addition)
````
